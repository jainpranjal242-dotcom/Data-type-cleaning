## Objective
Inspect incorrect data types and convert columns to appropriate types.

```python
import pandas as pd

df = pd.read_csv('employee_data_unclean.csv')
print('Original data types:')
print(df.dtypes)
df.head()
```

## 1. Convert employee ID to integer

```python
df['employee_id'] = pd.to_numeric(df['employee_id'], errors='raise').astype('int64')
print(df['employee_id'].dtype)
```

## 2. Convert salary to numeric

```python
df['salary'] = pd.to_numeric(df['salary'], errors='raise')
print(df['salary'].dtype)
```

## 3. Convert joining date to datetime

```python
df['joining_date'] = pd.to_datetime(df['joining_date'], errors='raise')
print(df['joining_date'].dtype)
```

## 4. Check corrected data types

```python
print(df.dtypes)
df.head()
```

## 5. Save cleaned dataset

```python
df.to_csv('employee_data_cleaned.csv', index=False)
print('Saved employee_data_cleaned.csv')
```

## Explanation of conversions
- `employee_id`: text to integer, since it contains whole-number IDs.
- `salary`: text to numeric, enabling calculations and statistics.
- `joining_date`: text to datetime, enabling date comparisons and calculations.
- `name` and `department`: remain text.

## Conclusion
Correct data types make sorting, filtering, calculations, and analysis more reliable.
